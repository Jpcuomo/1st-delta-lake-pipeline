```python
def saludar(nombre):
    return f"Hola {nombre}!"

# mi_funcion = saludar
# print(mi_funcion('Ana'))

def saludar_y_felicitar(fcion, nombre):
    resultado = fcion(nombre)
    return f'{resultado.title()} Excelente trabajo!!!'

funcion_decorada = saludar_y_felicitar(saludar, 'juan')
funcion_decorada
```

```python
@saludar_y_felicitar
def saludar(nombre):
    return f"Hola {nombre}!"

saludar('juan')
```

```python
# Paso 1: Función dentro de función
def conversacion():
    def decir_hola():
        return "Hola"
    
    def decir_adios():
        return "Adiós"
    
    print(decir_hola())
    print(decir_adios())

conversacion()
```

```python
def crear_procesador(tipo:str):
    def mayusculas(texto:str) -> str:
        return texto.upper()
    
    def minusculas(texto:str)-> str:
        return texto.lower()
    
    if tipo == 'mayuscula':
        return mayusculas
    else:
        return minusculas
    
texto = crear_procesador('mayuscula')('hola')
print(texto)
```
